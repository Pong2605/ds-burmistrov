# Перестановки с повторениями

```python
# Пример. Сколькими способами можно переставить буквы в слове «институт»?

# Решение. Здесь две буква «и», одна буква «н», 
# одна буква «с» , три буквы "т", одна буква «у», всего 8 букв. 
# Следовательно, число перестановок с повторениями равно

P_8_ = sp.factorial(8) / \
    (sp.factorial(2) * sp.factorial(1) * sp.factorial(1) * sp.factorial(3) * sp.factorial(1))
P_8_
```

```python
# Сколькими способами можно переставить буквы в слове «Миссисипи»?
from string import punctuation
word = 'Миссисипи'

def get_dct(word):  
    dct = {}
    for let in word.lower():
        dct[let] = dct.get(let, 0) + 1
    return dct

dct = get_dct(word)

def power(dct):
    pr = 1
    for i in dct.values():
        pr *= sp.factorial(i)
    return pr

def P_n_(dct):
    return sp.factorial(sum(dct.values())) / power(dct)

P_n_(dct)
```

```python
# Ребенку, не умеющему читать, дан набор карточек с буквами А, Б, Р, А, К, А, Д, А, Б, Р, А. 
# Сколькими способами ребенок может составить «слова», используя все карточки?

word = ''.join('А, Б, Р, А, К, А, Д, А, Б, Р, А'.split(', '))

dct = get_dct(word)

P_n_(dct)
```

```python
# Сколькими способами можно переставить буквы слова «решение», чтобы:

# а) три буквы «е» стояли рядом; б) слово начиналось с буквы «ш»?

word = 'решни'

dct = get_dct(word)

# a)
res = P_n_(dct)
print(res)

word = 'реение'

dct = get_dct(word)

# б)
res = P_n_(dct)
print(res)
```

# Сочетания без повторений

```python
def C_n_m(n, m):
    return sp.factorial(n) / (sp.factorial(m) * sp.factorial(n - m))
```

```python
C_n_m(20, 3)
```

```python
# На книжной полке стоят 10 различных книг по математике и 6 – по физике. 
# Сколькими способами можно выбрать две книги по одному предмету?

# Две книги по одному предмету можно выбрать способами:

C_n_m(6, 1) * C_n_m(10, 1)
```

# Сочетания с повторениями

```python
def C_n_m_(n, m):
    return sp.factorial(n + m -1) / (sp.factorial(m) * sp.factorial(n - 1))
```

```python
# а) В кондитерской имеется шесть видов пирожных. 
# Сколькими способами можно составить набор из трех пирожных?

res = C_n_m_(6, 3)
print(res)

# б) А если имеется три вида пирожных, а нужно составить набор из шести пирожных?

res = C_n_m_(3, 6)
print(res)
```

```python
# На прямой отмечено 5 точек, а на параллельной ей прямой – 6 точек.
# Сколько существует

# а) треугольников;

res = C_n_m(5, 1) * C_n_m(6, 2) + C_n_m(5, 2) * C_n_m(6, 1)
print(res)

# б) четырёхугольников с вершинами в этих точках

res = C_n_m(5, 2) * C_n_m(6, 2)
print(res)
```
